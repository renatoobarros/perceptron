## 引言

在动态变化的商业世界中，做出明智且具有战略性的决策是成功的关键。在这一背景下，感知机（Perceptron）作为一种受人脑工作机制启发的机器学习算法，能够成为数据分析与行为预测的强大工具。

感知机就像一个专注的学习者，通过观察数据中的模式与关联，学会做出预测并辅助决策。无论是识别具有销售潜力的商品、检测欺诈活动，还是个性化客户体验，它都能提供广泛的应用场景，推动当今企业的效率与智能化。

## 工作原理

感知机就像一个基本的人工神经网络：接收信息、处理信息并生成结果。该处理过程涉及几个简单的数学计算：

1. **线性组合：** 每条信息（以数值表示）乘以一个权重，该权重表示其在最终决策中的重要性。所有乘积相加后，得到一个称为"加权和"的中间值。

   公式：`u = (x1 * peso1) + (x2 * peso2) + ...`

2. **激活函数：** 上一步计算出的"加权和"(u) 传入激活函数，由激活函数决定感知机的最终输出。激活函数为模型引入非线性，使其能够学习数据之间的复杂关系。在本例中，我们使用"阶跃函数"（step function），其功能类似于一个开关：

   公式：

   `f(u) = 1, 若 u >= 0；0, 若 u < 0`

   其中：

   - `f(u)` 为激活函数的输出。
   - `u` 为上一步计算出的"加权和"。

   换言之，若"加权和"大于或等于零，感知机输出为 1；否则输出为 0。这种二值输出可以表示例如是否推荐某款商品、是否将客户归类为潜在买家，或任何需要预测的事件。

3. **Delta 函数：** 在使用激活函数计算出输出后，感知机将该输出与期望输出（即它本应预测的正确值）进行对比。期望输出与实际输出之间的差值称为"误差"。为了最小化该误差并从错误中"学习"，感知机使用 Delta 规则（即 Widrow-Hoff 规则）调整权重。该规则按误差与对应输入的比例调整每个权重。

   公式：
   `novo_peso = peso_atual + (taxa_de_aprendizagem * erro * entrada)`

   其中：

   - `taxa_de_aprendizagem` 是一个较小的值，控制学习速度（详见"参数与训练数据"部分）。

为了在实践中直观展示感知机的运作机制，以下 Python 代码使用商业场景中的示例和类比实现了一个简单模型。

## 参数与训练数据

正如实习生需要初始引导一样，感知机也需要初始参数和数据来启动其学习过程。

```python
# 感知机（Perceptron）神经网络实现
# Copyright (C) 2026 Renato Barros
#
# 本程序是自由软件：您可以根据自由软件基金会发布的 GNU 通用公共许可证
# 条款重新分发和/或修改它，无论是许可证的第 3 版，还是（由您选择）任何更高版本。
#
# 本程序的发布是希望它能发挥作用，但没有任何保证；甚至没有对适销性或特定用途适用性的
# 暗示保证。详情请参阅 GNU 通用公共许可证。
#
# 您应该已经随本程序收到了一份 GNU 通用公共许可证的副本。如果没有，请参阅
# <https://www.gnu.org/licenses/>。

# 初始化感知机权重（w1 和 w2）
peso = [0.2, -0.1]
# 设定算法的学习率
taxa_aprendizagem = 0.1
# 初始化感知机的总误差计数器为零
erros_totais = 0.0
# 设定训练的轮数（迭代）上限
limite_epocas = 20

# 定义训练数据及其对应的输入值（x1 和 x2）与期望输出（参考值）
dados_treinamento = [
    {"x1": 0.5, "x2": 0.8, "saida_desejada": 1},
    {"x1": 0.2, "x2": 0.4, "saida_desejada": 0},
]
```

- **初始化权重：** 在感知机中，"权重"代表每条信息在决策中的重要性。代码中 `peso = [0.2, -0.1]` 表示我们为两条相关信息（x1 和 x2）分配了随机初始权重。

  例如，假设我们正在构建一个预测商品是否能成为销售爆款的模型。此时 x1 可以是商品的好评数量，x2 是它被加入心愿单的次数。权重（0.2 和 -0.1）反映了这些信息在最终决策中的相对重要性。

- **设定学习率：** `taxa_aprendizagem = 0.1` 控制感知机在每次出错时调整权重的速度。可以将学习率类比为一架天平的灵敏度。

  较高的学习率意味着模型在每次出错时会对权重进行大幅调整，这可能加速学习，但也可能导致不稳定。较低的学习率则使模型进行小幅调整，过程更慢但更精确。

- **累计误差：** `erros_totais = 0.0` 充当误差计数器。在训练期间，感知机将预测值与真实值进行对比，每次误差都被计入。该计数器帮助我们评估模型的学习效果。

  例如，在一个训练感知机识别潜在买家的电商场景中，每当模型将用户错误分类为买家或非买家时，误差计数器就会递增，从而提供模型精确度的衡量。

- **限制训练轮数：** `limite_epocas = 20` 定义了感知机分析所有训练数据的最大次数。这是一种控制模型训练时间的方式。

  例如，一个学习根据卖家和买家位置、运输类型及配送历史来预测商品配送时间的感知机，每一轮训练就相当于分析所有已完成的配送记录，使模型能够调整权重以做出更精确的预测。

- **提供训练数据：** `dados_treinamento` 是感知机学习的基础。它是一个包含不同事件、商品或用户信息及其对应正确分类的数据集。

  在本代码中，我们有一个数据集（x1, x2）及其对应的结果标识（`saida_desejada = 1` 或 `0`）。

  假设我们正在训练感知机识别具有病毒式传播潜力的商品。训练数据将包含过去成功走红的商品信息，以及那些未能取得同样成功的商品。

## 感知机训练

在参数设定完毕、训练数据准备就绪后，感知机开始其学习过程。该过程就像一场微调博弈，感知机试图找到最佳的权重值，以便做出越来越精确的预测。

```python
# 启动感知机训练
for epoca in range(1, limite_epocas + 1):
    print(f"\n Época: {epoca}\n")
    # 标记当前轮是否存在误差，初始设为 False
    erro_na_epoca = False

    # 遍历每个训练数据
    for dado in dados_treinamento:
        # 提取当前数据的输入值和参考值
        x1 = dado["x1"]
        x2 = dado["x2"]
        saida_desejada = dado["saida_desejada"]

        # 显示当前权重值
        w1 = peso[0]
        w2 = peso[1]

        # 计算感知机输出 (u)
        u = (x1 * peso[0]) + (x2 * peso[1])
        # 应用激活函数（阶跃函数）：
        # 若 u >= 0，valor_saida = 1；否则 valor_saida = 0
        valor_saida = 1 if u >= 0 else 0

        # 计算误差：参考值与感知机输出之间的差值
        erro = saida_desejada - valor_saida

        # 累计总误差
        erros_totais += abs(erro)

        # 若当前迭代存在误差，将 erro_na_epoca 设为 True
        if erro != 0:
            erro_na_epoca = True
            # 使用 Delta 规则调整权重 (w1, w2)
            peso[0] += taxa_aprendizagem * erro * x1
            peso[1] += taxa_aprendizagem * erro * x2

        # 显示当前迭代信息（轮次、输入、输出、权重）
        print(
            f" X1: {x1:.1f}, X2: {x2:.1f} | W1: {w1:.2f}, W2: {w2:.2f} | Y: {saida_desejada}, "
            f"Y-predito: {valor_saida} | Erro: {erro}, Erro acumulado: {erros_totais}"
        )
        if erro_na_epoca:
            print(
                f" Ajuste de pesos: Novo W1 = {peso[0]:.2f}, Novo W2 = {peso[1]:.2f}"
            )
        if not erro_na_epoca:
            print(" Ajuste de pesos: Não foi necessário.\n")
    print("-=-" * 31)

    # 检查停止条件：若当前轮无误差（erro_na_epoca = False）
    if not erro_na_epoca:
        # 输出成功信息，指出达到零误差的轮次
        print(
            f"\n Modelo treinado com sucesso na época {epoca}! Erro zero atingido."
        )
        # 输出最终权重值 (w1, w2)
        print(
            f" Valores ideais de peso: W1 = {peso[0]:.2f}, W2 = {peso[1]:.2f}"
        )
        # 结束训练循环
        break
# 若循环达到最大轮数仍未成功
else:
    # 输出已达到最大轮数的信息
    print(
        f"\n O limite de {limite_epocas} épocas foi atingido sem alcançar erro zero."
    )
    # 提示感知机可能未能收敛
    print(" O perceptron não pôde ser treinado com sucesso.\n")
```

### 代码解析：

上述代码代表了感知机训练过程的核心，即一个在每一轮中重复的学习迭代循环，通过不断调整模型权重来最小化预测误差。让我们逐步分析每个环节：

1. **控制轮数：** 第一个 `for` 循环控制训练轮数，每一轮相当于一个完整的学习周期，感知机在其中分析所有训练数据、调整权重并计算总误差。之前设定的最大轮数 20 决定了这些周期的上限。

2. **应用训练数据：** 在每一轮中，第二个 `for` 循环逐一处理每条训练数据。就像感知机在逐个检查每个商品样本，学习其特征（x1, x2）及其正确分类（`saida_desejada`）。

3. **提取信息：** 对于训练数据中的每个商品，代码提取其特征（x1, x2）和正确分类（`saida_desejada`）。这些信息将用于计算感知机的输出，并随后调整权重。

4. **存储当前权重：** 变量 `w1` 和 `w2` 存储当前权重值。这样做是为了方便在训练过程中可视化权重变化。

5. **计算感知机输出：** 此处应用感知机的数学公式。"加权和"(u) 通过将商品特征（x1, x2）乘以其对应权重（w1, w2）后求和得出。在预测商品是否会走红的示例中，该加权和代表感知机对商品的"评估"，表明其成为销售爆款的倾向。

6. **应用激活函数：** "加权和"(u) 随后通过"阶跃"激活函数。该函数根据"加权和"决定感知机的输出（`valor_saida`）：若 u 大于或等于零，输出为 1（销售成功）；否则输出为 0（未成功）。

7. **计算误差：** 感知机将其预测（`valor_saida`）与正确分类（`saida_desejada`）进行对比。两者之间的差值代表模型的"误差"。正是基于误差，感知机调整权重，力求缩小预测值与真实值之间的差距。

8. **调整权重：** 若预测存在误差，则使用 Delta 规则调整权重（w1, w2）。该规则利用学习率控制调整幅度。调整的目标是使感知机在每次迭代中更接近正确分类。

9. **显示轮次信息：** 代码显示当前轮次的信息，包括交互中使用的输入和权重、期望输出、预测输出、误差、累计误差以及权重调整情况。这种详细的可视化使我们能够跟踪训练进度，理解权重在每一轮中如何被调整。

10. **检查停止条件：** 在每轮结束时，代码检查感知机是否已达到零误差，即是否正确分类了所有训练数据。若达到零误差，训练成功终止，并显示模型的最终权重。否则，训练继续直到达到最大轮数。

感知机的训练过程旨在找到一个平衡点，通过调整权重使误差最小化。这一学习过程虽然相对简单，却是开发更复杂、更强大的人工智能模型的基础。

## 结果解读

分析模型运行后生成的结果，使我们能够评估感知机的学习过程并判断训练是否成功。训练成功的定义是模型能够正确分类所有训练数据，在本例中即达到零误差。

观察结果可以发现，感知机在第 12 轮达到了零误差。这意味着模型需要 12 次遍历训练数据才能找到一组能够正确分类数据的权重。

需要多轮才能达到零误差意味着，尽管训练数据可以被一条线分为两类（线性可分），但对于训练初期接收到的随机权重而言，这种分离并不是一目了然的。因此，初始权重不足以正确分类数据，需要通过多轮逐步调整。

在每一轮中，感知机根据分类误差调整权重，力求缩小预测值与真实值之间的距离。每次调整后，模型都更接近理想解，直到第 12 轮所有数据都被正确分类。

最终得到的权重 `W1 = 0.23` 和 `W2 = -0.14` 反映了模型在训练过程中获得的知识。分析这些权重使我们能够理解感知机如何学会分类数据：

- **W1 为正 (0.23)：** 表明特征 x1 对模型的最终决策具有正向影响。x1 的值越大，感知机越倾向于输出"1"，即根据具体问题，可能代表推荐某款商品、将客户归类为潜在买家，或任何需要预测的事件。

- **W2 为负 (-0.14)：** 表明特征 x2 对最终决策具有负向影响。与 x1 不同，x2 的值越大，模型越不倾向于输出"1"，起到与 x1 相互制衡的作用。

这种权重分析展示了感知机权衡不同特征并学习数据之间复杂关系的能力。

需要特别指出的是，若最大轮数设为 3，训练将在感知机达到零误差之前被中断。仅 3 轮训练，模型没有足够的时间调整权重以正确分类所有训练数据。在这种场景下，最终权重将代表一个次优解，损害模型的泛化能力，即其在训练期间未见过的数据上表现良好的能力。
